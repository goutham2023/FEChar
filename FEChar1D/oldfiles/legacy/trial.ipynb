```python
import numpy as np

# ---- Load data ----
data = np.loadtxt("pressure_pyrogas.dat")

P = data[:, 0]
T = data[:, 1]
Mu = data[:, 2]
cp = data[:, 3]
gamma = data[:, 4]
H = data[:, 5]

# ---- Get unique sorted values ----
Pvals = np.unique(P)
Tvals = np.unique(T)

nP = len(Pvals)
nT = len(Tvals)

print("Unique P:", nP)
print("Unique T:", nT)

# ---- Initialize grids ----
Mu_grid = np.zeros((nT, nP))
cp_grid = np.zeros((nT, nP))
gamma_grid = np.zeros((nT, nP))
H_grid = np.zeros((nT, nP))

# ---- Fill grids ----
# Map (P,T) → index
for i in range(len(P)):
    ip = np.where(Pvals == P[i])[0][0]
    it = np.where(Tvals == T[i])[0][0]

    Mu_grid[it, ip] = Mu[i]
    cp_grid[it, ip] = cp[i]
    gamma_grid[it, ip] = gamma[i]
    H_grid[it, ip] = H[i]

# ---- Save to file ----
with open("grid_data.dat", "w") as f:
    # sizes
    f.write(f"{nP} {nT}\n")

    # P values
    f.write(" ".join(map(str, Pvals)) + "\n")

    # T values
    f.write(" ".join(map(str, Tvals)) + "\n")

    # Function to write grid
    def write_grid(grid, name):
        f.write(f"# {name}\n")
        for i in range(nT):
            f.write(" ".join(map(str, grid[i, :])) + "\n")

    write_grid(Mu_grid, "Mu")
    write_grid(cp_grid, "cp")
    write_grid(gamma_grid, "gamma")
    write_grid(H_grid, "H")

print("Grid file written: grid_data.dat")
```
